```python
import os, requests
from datetime import datetime
from zoneinfo import ZoneInfo

VOLUME = "/Volumes/workspace/heatcheck/raw"
ENDPOINTS = {
    "wbgt": "https://api-open.data.gov.sg/v2/real-time/api/weather?api=wbgt",
    "air_temperature": "https://api-open.data.gov.sg/v2/real-time/api/air-temperature",
}

now = datetime.now(ZoneInfo("Asia/Singapore")).strftime("%Y%m%d_%H%M%S")

for name, url in ENDPOINTS.items():
    r = requests.get(url, timeout=15)
    r.raise_for_status()
    folder = f"{VOLUME}/{name}"
    os.makedirs(folder, exist_ok=True)
    path = f"{folder}/{name}_{now}.json"
    with open(path, "w") as f:
        f.write(r.text)
    print("Saved", path)
```
