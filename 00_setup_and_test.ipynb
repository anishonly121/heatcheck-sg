```python
import requests

tests = {
    "Air temperature": "https://api-open.data.gov.sg/v2/real-time/api/air-temperature",
    "WBGT heat stress": "https://api-open.data.gov.sg/v2/real-time/api/weather?api=wbgt",
    "HDB dataset": "https://data.gov.sg/api/action/datastore_search?resource_id=d_17f5382f26140b1fdae0ba2ef6239d2f&limit=1",
}

for name, url in tests.items():
    try:
        r = requests.get(url, timeout=15)
        print(f"{name}: status {r.status_code}")
        print(r.text[:300], "\n")
    except Exception as e:
        print(f"{name}: FAILED -> {type(e).__name__}: {e}\n")
```
